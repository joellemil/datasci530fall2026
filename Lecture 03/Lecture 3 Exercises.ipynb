<font size = "3">

- Import the dataset "data_raw/results.csv" to a new object "results"

to missing values (NaNs) and covert to numeric
- To do so, go to Lecture 13 and follow the setups under the headings

    - "Replace certain values"

HINT: substitute "milliseconds" instead of "alt", <br>

````python
# Write your own code

results = pd.read_csv('data_raw/results.csv')

# From 'Extract list of non-numeric values'

subset = results.query('milliseconds.str.isnumeric() == False')
list_unique = pd.unique(subset["milliseconds"])
print(list_unique)

# This command replaces the values of the "milliseconds" column
results["milliseconds"] = results["milliseconds"].replace('\\N', np.nan)

# pd.to_numeric() converts a column to numeric
# Before you use this option, make sure to "clean" the variable
# as we did before by checking what the non-numeric values are

results["milliseconds"] = pd.to_numeric(results["milliseconds"])

````

(g) Recode a numeric colum

with the following categories

$\qquad$ ``` ["1950-1970","1971-1990","1991-2005", "2006-2020", "2020-onwards"] ```

$\qquad$ HINT: Use the "pd.cut()" command. 

````python
# Write your own code

races = pd.read_csv('data_raw/races.csv')

bins_years = [1950, 1970, 1990, 2005, 2020, float('inf')]
labels_years = ['1950-1970', '1971-1990', '1991-2005', '2006-2020', '2020-onwards']

races['year_brackets'] = pd.cut(
    races['year'],
    bins=bins_years,
    labels=labels_years,
    right=True
)

races
````

(h) Aggregate and query

<font size = "3">

- Import the "data_raw/constructor_standings.csv" file
- Subset the observations for 'raceId == 10' using ```.query()```
- Obtain the sum of "points" by "constructorId" <br> 
(the team) using ``` .groupby().agg() ```
- Subset the teams with a sum of points $\ge 50$
- Store the results of all these operations in a dataframe <br>
```constructor_standings_agg ```

````python
# Write your own code

constructor_standings = pd.read_csv('data_raw/constructor_standings.csv')

constructor_standings_agg = (
    constructor_standings.query("raceId == 10")
    .groupby('constructorId')
    .agg(sum_points = ('points', 'sum'))
    .query("sum_points > 50")
)

constructor_standings_agg


````

(i) Aggregate and sort

<font size = "3">

- Import the "data_raw/constructor_standings.csv" file
- Obtain the sum of "points" by "constructorId" <br> 
(the team) using ``` .groupby().agg() ```
- Sort the teams in descending order using of <br>
the sum of points using ``` .sort_values() ```

````python
# Write your own code

constructor_standings = pd.read_csv('data_raw/constructor_standings.csv')

constructor_standings_agg = (
    constructor_standings
    .groupby('constructorId')
    .agg(sum_points = ('points', 'sum'))
    .sort_values('sum_points', ascending=False)
)

constructor_standings_agg



````

(k) Merge dataset

<font size = "3">

- Import the "data_raw/constructor_standings.csv" and <br>
"data_raw/constructors.csv" files
- Create a new dataset with  ```pd.merge()``` using <br>
"constructor_standings" as the primary dataset, and <br>
"constructors" as the secondary dataset, merging on <br>
the column "constructorId"
- To get full points **ONLY** merge the "nationality" column <br>
from the secondary dataset (not all the columns)

HINT: Use ```[[...]]``` to extract a subset of columns <br>
from the secondary dataset before merging, including the <br>
"constructorId" column

````python
# Write your own code

constructor_standings = pd.read_csv('data_raw/constructor_standings.csv')
constructors = pd.read_csv('data_raw/constructors.csv')

constructors_merge = pd.merge(constructor_standings,
                       constructors[['constructorId', 'nationality']],
                       on = "constructorId",
                       how = "left")

constructors_merge


````

<font size = "5">

(l) Formatted results table

<font size = "3">

-  For the dataset you computed above construct a \
descriptive statistics table
- Round the numbers and change the column names to something \
more interpretable

````python
# Write your own code

constructors_merge = constructor_standings.merge(
    constructors[['constructorId', 'name', 'nationality']],
    on='constructorId'
)

constructors_merge[['name', 'nationality', 'points', 'position', 'wins']].rename(columns={
    'name': 'Manufacturer',
    'nationality': 'Nationality',
    'points': 'Total Championship Points',
    'position': 'Championship Position',
    'wins': 'Race Wins'
}).style.format({
    'Points' : '{:,.1f}'
}).set_caption("Manufacturer Performance Summary").hide(axis="index")

#QUESTION - is this the type of descriptive statistics expected or should it be like below?

manufacturer_stats = (
    constructors_merge
    .groupby("name")
    .agg(
        total_points=("points", "sum"),
        average_position=("position", "mean"),
        total_wins=("wins", "sum")
    )
    .sort_values("total_points", ascending=False)
    .reset_index()
)

manufacturer_stats


manufacturer_stats.rename(columns={
    "name": "Manufacturer",
    "total_points": "Total Points",
    "average_position": "Average Position",
    "total_wins": "Total Wins"
}).style.format({
    "Total Points": "{:,.0f}",
    "Average Position": "{:,.1f}",
    "Total Wins": "{:,.0f}"
}).set_caption("Manufacturer Performance Summary").hide(axis="index")

````
